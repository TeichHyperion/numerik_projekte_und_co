```python
import math
from numpy import sign
from decimal import Decimal

# Non linear Polarizability equation 
def polarzation_dens(chi, d , E_field ,chi_3):
    """ d in C/V**2 and E_field in W/cm**2 """
    e_0         = 8.85418782 * 10**-12
    P_linear, P_quadratic, P_cubic = 0
    if chi != 0:
        P_linear    = e_0 * chi * E_field 
    if d !=0:
        P_quadratic = 2 * d *E_field**2 
    if chi_3 != 0:
        P_cubic     = 4 * chi_3 * E_field**3
    P               = P_linear + P_quadratic + P_cubic
    return  P, P_linear, P_quadratic, P_cubic
    
def intensity(E_field, eta):
    I = abs(E_field)**2 / (2*eta)
    return I 

def E_field(intensity, eta):
    E_field_val = math.sqrt(intensity * 2 * eta)
    return E_field_val

def eta(n):
    """eta (impedence) in Ohm. Numerical Value for eta_0 taken from 'fundamentals of photonics'. n is the refractive index  """
    eta_0 = 377 #= math.sqrt((mu_0/epsilon_0))
    eta = eta_0 / n
    return eta
```

```python
eta_val         = eta(n=1.5)
E_field_val     = E_field(intensity=10**10, eta=eta_val)
polarzation_val = polarzation_dens(chi=0, d=6.8*10**-21, E_field=E_field_val )
print(polarzation_val)
```

```python
# basic bisection solver.

def bisection(func, lower_x, upper_x, max_iter=10**6):
    """Assumes on sign change in the range"""
    #current_val = func(start)
    #initial setup

    lower_val = func(lower_x)
    upper_val = func(upper_x)
    if sign(lower_val) == sign(upper_val):
        return lower_val, upper_val, "Error, no Zeros (or even number) in this range. \n"

    for _ in range(max_iter):
        midpoint_x = (upper_x + lower_x)/2
        midpoint_val  = func(midpoint_x)

        if abs(midpoint_val) < 10**-6 or abs(lower_x - upper_x) < 10**-6:
            
            return midpoint_x, midpoint_val

        if sign(midpoint_val) == sign(upper_val):
            upper_val = midpoint_val
            upper_x   = midpoint_x
            #new_val = new_val
        else:   
            #old_val = old_val
            lower_val = midpoint_val
            lower_x   = midpoint_x

```

```python
root = bisection(lambda x: 1.5**2-1-x, 0, 10)
print(root)
```
